```python
import csv
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np

from utils import graphics
from buryn_utils import voltage_utils

graphics.set_matplotlib_settings()
```

```python
BURYNOUTPUTS_DIR = "../data/buryn_data/buryn_outputs_11"

steps, voltages, neuron_ids = voltage_utils.load_voltage_recording(directory=f"{BURYNOUTPUTS_DIR}/recordings", pattern="voltages_step*.txt")
print(f"Loaded {len(steps)} timesteps for {len(neuron_ids)} neuron(s)")
```

```python
plt.figure(figsize=(20, 10))
for idx, neuron_id in enumerate(neuron_ids):
    plt.plot(steps, voltages[:, idx], label=f"Neuron {neuron_id}")
plt.xlabel("Timestep")
plt.ylabel("Membrane\n potential (V)")
plt.title("Voltages of Neurons Over Time")
plt.grid(True, linestyle='--', alpha=0.3)
plt.axvspan(5000, 6000, color='yellow', alpha=0.3, label='Stimulation of 1029 (10ms)')
plt.ylim(-0.07, -0.035)
plt.xlim(0, 100000)
plt.legend(loc='upper left')
plt.tight_layout()
plt.savefig(BURYNOUTPUTS_DIR+"/plots/voltages.png", dpi=300)
plt.show()
```
